```python
def play_game():
    import random as rnd

    sym = {"rock": 1, "paper": 1, "scissors": 1}
    player_hp = 8
    moves = ["rock", "paper", "scissors"]

    # 1. Outer loop for 3 bossess
    for boss_num in range(1, 4):
        boss_hp = 5  # Reset boss HP at start of each encounter
        print(
            f"\n=== ENCOUNTER {boss_num}: BOSS HP = {boss_hp} | PLAYER HP = {player_hp} ==="
        )

        # 2. Combat loop for a single boss
        while player_hp > 0 and boss_hp > 0:
            boss_symbol = rnd.choice(moves)  # Boss picks first

            # Player input validation loop
            while True:
                used_symbol = input("Make your choice (r/p/s): ").lower().strip()
                if used_symbol in ["r", "rock"]:
                    used_symbol = "rock"
                    break
                elif used_symbol in ["p", "paper"]:
                    used_symbol = "paper"
                    break
                elif used_symbol in ["s", "scissors"]:
                    used_symbol = "scissors"
                    break
                else:
                    print("Invalid choice! Please try again.\n")

            # Reveal boss move & compare combat
            print(f"Boss chose: {boss_symbol}")
            if used_symbol == boss_symbol:
                print("It's a tie! No damage taken.")
            elif (
                (used_symbol == "rock" and boss_symbol == "scissors")
                or (used_symbol == "paper" and boss_symbol == "rock")
                or (used_symbol == "scissors" and boss_symbol == "paper")
            ):
                damage_dealt = sym[used_symbol]
                boss_hp -= damage_dealt
                print(
                    f"You win the round! You deal {damage_dealt} damage to the Boss."
                )
            else:
                player_hp -= 1
                print("Boss wins the round! You take 1 damage.")

            print(f"Current Status -> Player HP: {player_hp} | Boss HP: {boss_hp}")

        # 3. Post-fight check (After combat loop finishes)
        if player_hp > 0:
            print(f"\nVictory! You defeated Boss {boss_num}!")

            # Upgrade symbol if it's not the final boss
            if boss_num < 3:
                print(f"Current Symbol Damage: {sym}")
                while True:
                    upgrade_choice = (
                        input(
                            "Which symbol do you want to upgrade by +1 damage? (rock/paper/scissors): "
                        )
                        .lower()
                        .strip()
                    )

                    if upgrade_choice in ["r", "rock"]:
                        upgrade_choice = "rock"
                    elif upgrade_choice in ["p", "paper"]:
                        upgrade_choice = "paper"
                    elif upgrade_choice in ["s", "scissors"]:
                        upgrade_choice = "scissors"

                    if upgrade_choice in sym:
                        sym[upgrade_choice] += 1
                        print(
                            f"Upgraded! {upgrade_choice.capitalize()} now deals {sym[upgrade_choice]} damage.\n"
                        )
                        break
                    else:
                        print(
                            "Invalid symbol! Please choose rock, paper, or scissors."
                        )
        else:
            print("\nGame Over! You were defeated.")
            break  # Exit boss loop if player dies
```

```python
play_game()
```
